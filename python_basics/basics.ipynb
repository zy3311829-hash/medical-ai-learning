优化循环

```python
xj=0
while xj<100:
    mj=0
    while mj<34:
        gj=0
        while gj<20:
            if (mj*3+5*gj+xj/3==100)&(mj+gj+xj==100):
                print(f"公鸡数量为{gj},母鸡数量为{mj},小鸡数量为{xj}")
                break
            else:
                gj+=1
        mj+=1
    xj+=1
```

```python
gj=0
while gj<20:
    mj=0
    while mj<34:
        xj=0
        while xj<100:
            if (mj*3+5*gj+xj/3==100)&(mj+gj+xj==100):
                print(f"公鸡数量为{gj},母鸡数量为{mj},小鸡数量为{xj}")
                break
            else:
                xj+=1
        mj+=1
    gj+=1
```

上面的这两个例子很有意思。打个比方，好比先划分建筑，再划分楼层，再划分房间一样，我们的最终目的是找到对应的房间。如果房间数恒定，我肯定希望楼数越少越好，这样我开一个楼就可以循环检查更多的房间。

```python
gj=0
while gj<20:
    mj=0
    while mj<34:
        if(gj*5+mj*3+(100-gj-mj)/3==100):
            print(f"公鸡数量为{gj},母鸡数量为{mj},小鸡数量为{100-gj-mj}")
            break
        else:
            mj+=1
    gj+=1
```

self不是python的关键字，也不是类天生自带的东西，它只是一个普通的函数参数，只不过大家约定俗成的把它叫做self

这两个代码只差了两行，却能够解决我困惑了非常久的问题，真的是太好了！

```python
class Jzt:#相当于定义了一个类
    shape="*"#这些是属性,定义的时候会给默认值，用的时候可以修改
    row=5

    def draw(self,row,shape):#这些是动
        self.shape=shape
        self.row=row
        for i in range(row):
            for j in range(-1,i):
                print(shape,end="")
            print()
p=Jzt()
shape="%"#a="%"
row=4#b=4
p.draw(row,shape)#p.draw(a,b)
print(shape,row)

#看得见摸得着的叫做对象，对象之间有一些共同性质，也有一些不同性质
#把看的见的东西（对象）相似的东西！抽象！出来，变成类，类看不见不存在
#python面向对象 C语言面向过程
#类有属性（对人来说吃饭睡觉），类也有动？
```

```python
class Dog:
    name="luke"#这个name是类的属性
    def woof(self):

        print(f"{self.name}:wangwang")#这里的self.name的name是类给这个self的属性
a=Dog()
a.woof()
```

2.是这个对象的属性

```python
class Dog():
    name="luck"#这个类的属性
    def woof(self):

        print(f"{self.name}:wangwang")
a=Dog()
a.name="flora"#这个地方更新了对象的属性
a.woof()#这个括号里面只要不是空的就行""这个都可以，因为需要给name一个输入，或者定义函数的时候就不要定义name的输入，这里也不用放东西了
print(Dog.name)
```

3.只是一个传输进去的变量，无法彻底更改对象或者类的属性

```python
class Dog:
    name="luke"
    def woof(self,name):

        print(f"{name}:wangwang")

a=Dog()
name="flora"
a.woof(name)
print(a.name)
print(Dog().name)
```
