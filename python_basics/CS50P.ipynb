### Math Interpreter

```python
x, y, z = input("Expression: ").split()

x = int(x)
z = int(z)

if y == "+":
    result = x + z
elif y == "-":
    result = x - z
elif y == "*":
    result = x * z
elif y == "/":
    result = x / z

print(f"{result:.1f}")
```

### Meal Time


```python
def main():
    time = input("What time is it? ")
    time = convert(time)

    if 7 <= time <= 8:
        print("breakfast time")
    elif 12 <= time <= 13:
        print("lunch time")
    elif 18 <= time <= 19:
        print("dinner time")


def convert(time):
    hour, minute = time.split(":")
    hour = int(hour)
    minute = int(minute)

    return hour + minute / 60
```

### Vanity Plates

```python
def main():
    plate = input("Plate: ")

    if is_valid(plate):
        print("Valid")
    else:
        print("Invalid")


def is_valid(s):
    if len(s) < 2 or len(s) > 6:
        return False

    if not s[0:2].isalpha():
        return False

    if not s.isalnum():
        return False

    number_started = False

    for c in s:
        if c.isdigit():
            if number_started == False:
                if c == "0":
                    return False

                number_started = True

        else:
            if number_started:
                return False

    return True


main()
```

### Fuel Gauge


```python
while True:
    try:
        fraction = input("Fraction: ")

        x, y = fraction.split("/")

        x = int(x)
        y = int(y)

        if x > y:
            continue

        percentage = round(x / y * 100)

        break

    except (ValueError, ZeroDivisionError):
        pass


if percentage <= 1:
    print("E")
elif percentage >= 99:
    print("F")
else:
    print(f"{percentage}%")
```

### Frank,lan and Glen's Letters
程序文件名本身也算一个命令行参数

```python
import sys
import random
from pyfiglet import Figlet

figlet = Figlet()

if len(sys.argv) == 1:
    fonts = figlet.getFonts()
    figlet.setFont(font=random.choice(fonts))

elif len(sys.argv) == 3 and sys.argv[1] in ["-f", "--font"]:
    fonts = figlet.getFonts()

    if sys.argv[2] not in fonts:
        sys.exit("Invalid usage")

    figlet.setFont(font=sys.argv[2])

else:
    sys.exit("Invalid usage")

text = input("Input: ")
print("Output:")
print(figlet.renderText(text))
```

python文件路径写法
from pathlib import Path（从自带的工具箱里面拿出来path这个工具）
BASE_DIR = Path(__file__).resolve().parent
__file__表示正在运行的这个python文件在哪里（__xxx__不是普通的变量名，往往是自动调用用的）
Path(__file__)原来的路径就只是一串文字，但经过Path函数，python知道了这是一个路径（知道找父级，检查是否存在等...）
.resolve是为了得到更加明确的绝对路径，有时候得到的是.\train.py（.表示当前目录），经过.resolve之后的得到更见明确的D:\image_classification\train.py
.parent表示这个文件的上一级文件夹
所以BASE_DIR = D:\AI\image_classification
mkdir的意思就是make directory
